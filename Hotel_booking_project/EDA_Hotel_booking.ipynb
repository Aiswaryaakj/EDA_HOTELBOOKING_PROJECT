# Hotel Booking Reservation Analysis

## Introduction


In the hospitality industry, analyzing customer booking behavior plays a vital role in optimizing revenue, enhancing guest experience, and minimizing cancellations. Hotels face constant challenges in managing fluctuating demand, predicting customer preferences, and ensuring efficient resource allocation.This project focuses on exploring hotel booking data to identify meaningful patterns related to customer behavior, seasonal trends, and booking dynamics. The dataset includes key variables such as booking dates, length of stay, customer segments, average daily rate (ADR), and special requests made by guests.

By applying data analysis techniques, this study aims to uncover insights that can support better decision-making, improve operational efficiency, and reduce cancellation rates. Understanding these patterns allows hotels to implement targeted strategies for pricing, marketing, and customer engagement.

What is Hotel Booking ?


Hotel booking is the process of reserving a room or accommodation in a hotel for a specific period of time. Customers book hotels for travel, vacations, business trips, events, or other purposes. The booking can be made through hotel websites, travel agencies, mobile apps, or online platforms.

A hotel booking usually includes details such as:

Customer information    
Check-in and check-out dates        
Room type           
Number of guests                    
Payment details             
Booking status (confirmed, canceled, etc.)

In data analysis projects, hotel booking data is used to study customer behavior, booking trends, pricing, seasonal demand, and cancellation patterns to help hotels improve their services and business decisions.


Key Variables in the Hotel Booking Dataset :

* hotel — Type of hotel booked
* is_canceled — Indicates whether the booking was canceled
* lead_time — Number of days between booking and arrival
* arrival_date_month — Month of customer arrival
* stays_in_weekend_nights — Number of weekend nights stayed
* stays_in_week_nights — Number of weekday nights stayed
* adults — Number of adults included in the booking
* children — Number of children included in the booking
* meal — Type of meal plan selected
* country — Country of the customer
* market_segment — Booking market segment category
* distribution_channel — Method used to distribute the booking
* is_repeated_guest — Indicates whether the customer is a repeated guest
* previous_cancellations — Number of previous canceled bookings
* reserved_room_type — Type of room reserved by the customer
* assigned_room_type — Type of room assigned at check-in
* booking_changes — Number of booking modifications made
* deposit_type — Deposit payment type for the booking
* customer_type — Category of customer
* adr — Average Daily Rate of the booking
* required_car_parking_spaces — Number of parking spaces requested
* total_of_special_requests — Number of special requests made by the customer
* reservation_status — Final reservation status
* reservation_status_date — Date of reservation status update

## Q1.Which customer type has the highest number of hotel bookings ?

```python
customer_counts = df["customer_type"].value_counts()

print("Customer Type Counts:\n", customer_counts)

plt.figure(figsize=(8,5))
bars = plt.bar(customer_counts.index, customer_counts.values)

plt.title("Customer Type Distribution Analysis")
plt.xlabel("Customer Type")
plt.ylabel("Number of Bookings")
plt.xticks(rotation=45)

plt.show()
```

Transient customers have the highest number of hotel bookings, while Group customers have the lowest bookings. This shows that most hotel bookings are made by short-term individual travelers.

 ## Q2.What percentage of bookings are canceled ?

```python
cancel_counts = df["is_canceled"].value_counts()

labels = ["Not Canceled", "Canceled"]

plt.figure(figsize=(6,6))

plt.pie(
    cancel_counts.values,
    labels=labels,
    autopct='%1.1f%%',
    startangle=90
)

plt.title("Booking Cancellation Analysis")
plt.show()
```

A significant percentage of hotel bookings are canceled.
